experimenting with quantum states in simple BB84 test

```python
import math

zero_probability = abs(1/math.sqrt(2))**2
one_probability = abs(1/math.sqrt(2))**2

print(f"P(0): {zero_probability}")
print(f"P(1): {one_probability}")
```

```python
import math

SQRT_HALF = 1 / math.sqrt(2)

ZERO = (1, 0)
ONE = (0, 1)

PLUS = (SQRT_HALF, SQRT_HALF)
MINUS = (SQRT_HALF, -SQRT_HALF)

print("|0> =", ZERO)
print("|1> =", ONE)
print("|+> =", PLUS)
print("|-> =", MINUS)

def probabilities(state):
    return tuple(abs(amplitude) ** 2 for amplitude in state)

print("Probabilities for |0>:", probabilities(ZERO))
print("Probabilities for |1>:", probabilities(ONE))
print("Probabilities for |+>:", probabilities(PLUS))
print("Probabilities for |->:", probabilities(MINUS))
```
